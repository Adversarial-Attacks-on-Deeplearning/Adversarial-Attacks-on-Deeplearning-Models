```python
import os
import cv2
from tqdm import tqdm

# Define paths
original_path = 'GTSRB/Final_Training/Images'
output_path = 'GTSRB_processed'

# Create output directory if it doesn't exist
os.makedirs(output_path, exist_ok=True)

# Process each class folder
for class_folder in tqdm(sorted(os.listdir(original_path))):
    class_path = os.path.join(original_path, class_folder)

    # Skip if not a directory
    if not os.path.isdir(class_path):
        continue

    # Get all image files in the class folder
    for image_file in os.listdir(class_path):
        # Skip non-image files (like .csv files)
        if not image_file.lower().endswith(('.ppm', '.png', '.jpg', '.jpeg')):
            continue

        # Read the image
        img_path = os.path.join(class_path, image_file)
        img = cv2.imread(img_path)

        # Create new filename: label_adv_index.png
        # Remove file extension and any numbers after it (some files have .ppm.1)
        base_name = os.path.splitext(image_file)[0].split('.')[0]
        new_filename = f"{class_folder}_adv_{base_name}.png"

        # Save as PNG
        output_file = os.path.join(output_path, new_filename)
        cv2.imwrite(output_file, img)

print("Processing complete! All images have been converted to PNG and renamed.")
```

```python
import os
import shutil

# Define your paths (replace with your actual paths)
folder1_path = '/content/UAP_train10'  # First folder path
folder2_path = '/content/UAP_train5'  # Second folder path
merged_path = '/content/UAP_merged'    # Destination folder

# Create the merged folder if it doesn't exist
os.makedirs(merged_path, exist_ok=True)

def merge_folders(source1, source2, destination):
    # First copy all files from folder1
    for filename in os.listdir(source1):
        src_file = os.path.join(source1, filename)
        dst_file = os.path.join(destination, filename)
        shutil.copy2(src_file, dst_file)

    # Then copy files from folder2, appending '_delta5' to duplicates
    for filename in os.listdir(source2):
        src_file = os.path.join(source2, filename)
        dst_file = os.path.join(destination, filename)

        # If file exists, add '_delta5' before extension
        if os.path.exists(dst_file):
            name, ext = os.path.splitext(filename)
            new_filename = f"{name}_delta5{ext}"  # e.g., "cat.jpg" → "cat_delta5.jpg"
            dst_file = os.path.join(destination, new_filename)

        shutil.copy2(src_file, dst_file)

# Merge the folders
merge_folders(folder1_path, folder2_path, merged_path)

print(f"Merged {len(os.listdir(folder1_path)) + len(os.listdir(folder2_path))} files")
print(f"Result stored in: {merged_path}")
```

```python
import os
import pandas as pd
from PIL import Image

def build_dataset_csv():
    data = []

    # Process All_fgsm_pgd
    for filename in os.listdir("All_fgsm_pgd"):
        if filename.endswith(".png"):
            label = int(filename.split("_")[2])
            data.append({"image_path": f"All_fgsm_pgd/{filename}", "label": label, "source": "fgsm"})

    # Process UAP_merged
    for filename in os.listdir("UAP_merged"):
        if filename.endswith(".png"):
            label = int(filename.split("_")[0])
            data.append({"image_path": f"UAP_merged/{filename}", "label": label, "source": "uap"})

    # Process GTSRB (convert .ppm to .png if needed)
    for label_dir in os.listdir("GTSRB/Final_Training/Images"):
        label = int(label_dir)
        for img_file in os.listdir(f"GTSRB/Final_Training/Images/{label_dir}"):
            if img_file.endswith(".ppm"):
                # Convert to PNG (optional)
                img_path = f"GTSRB/Final_Training/Images/{label_dir}/{img_file}"
                png_path = img_path.replace(".ppm", ".png")
                Image.open(img_path).save(png_path)  # Save as PNG
                data.append({"image_path": png_path, "label": label, "source": "gtsrb"})

    # Save to CSV
    df = pd.DataFrame(data)
    df.to_csv("dataset.csv", index=False)
    return df

build_dataset_csv()
```

```python
import tensorflow as tf
import pandas as pd
from sklearn.model_selection import train_test_split
# Load CSV (columns: image_path, label)
df = pd.read_csv("dataset.csv")
train_df, val_df = train_test_split(df, test_size=0.2, random_state=42)

# Define data pipeline (no normalization)
def load_image(path, label):
    image = tf.io.read_file(path)
    image = tf.image.decode_png(image, channels=3)  # Values in [0, 255]
    image = tf.image.resize(image, [240, 240])      # Resize to 240x240
    return image, label

# Create datasets
train_ds = tf.data.Dataset.from_tensor_slices((train_df["image_path"], train_df["label"]))
train_ds = train_ds.map(load_image).batch(32).prefetch(tf.data.AUTOTUNE)

val_ds = tf.data.Dataset.from_tensor_slices((val_df["image_path"], val_df["label"]))
val_ds = val_ds.map(load_image).batch(32).prefetch(tf.data.AUTOTUNE)
model = tf.keras.models.load_model("/content/EfficientNetB1.keras")
print(model.input_shape)  # e.g., (None, 224, 224, 3)
input_layer = tf.keras.layers.Input(shape=(240, 240, 3))
x = tf.keras.layers.Resizing(model.input_shape[1], model.input_shape[2])(input_layer)  # Resize to original
x = model(x)  # Pass through original model (pixels remain [0, 255])
new_model = tf.keras.Model(inputs=input_layer, outputs=x)
new_model.compile(
    optimizer=tf.keras.optimizers.Adam(1e-4),
    loss="sparse_categorical_crossentropy",  # For integer labels
    metrics=["accuracy"]
)
history = new_model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=10
)
```

```python
import os
import shutil
import pandas as pd
from PIL import Image

def build_dataset_csv():
    target_folder = "/content/All_Adv_plus_Training_samples"  # Specified target folder name
    os.makedirs(target_folder, exist_ok=True)  # Create folder if it doesn't exist

    data = []
    counter = 1  # Counter for unique filenames

    # Process All_fgsm_pgd
    for filename in os.listdir("All_fgsm_pgd"):
        if filename.endswith(".png"):
            # Create new filename and copy image
            new_filename = f"img_{counter}.png"
            shutil.copy2(f"All_fgsm_pgd/{filename}", f"{target_folder}/{new_filename}")

            label = int(filename.split("_")[2])
            data.append({
                "image_path": f"{target_folder}/{new_filename}",
                "label": label,
                "source": "fgsm",
                "original_filename": filename
            })
            counter += 1

    # Process UAP_merged
    for filename in os.listdir("UAP_merged"):
        if filename.endswith(".png"):
            # Create new filename and copy image
            new_filename = f"img_{counter}.png"
            shutil.copy2(f"UAP_merged/{filename}", f"{target_folder}/{new_filename}")

            label = int(filename.split("_")[0])
            data.append({
                "image_path": f"{target_folder}/{new_filename}",
                "label": label,
                "source": "uap",
                "original_filename": filename
            })
            counter += 1

    # Process GTSRB (convert .ppm to .png)
    for label_dir in os.listdir("GTSRB/Final_Training/Images"):
        label = int(label_dir)
        for img_file in os.listdir(f"GTSRB/Final_Training/Images/{label_dir}"):
            if img_file.endswith(".ppm"):
                # Convert to PNG and save in target folder
                new_filename = f"img_{counter}.png"
                img_path = f"GTSRB/Final_Training/Images/{label_dir}/{img_file}"
                Image.open(img_path).save(f"{target_folder}/{new_filename}")

                data.append({
                    "image_path": f"{target_folder}/{new_filename}",
                    "label": label,
                    "source": "gtsrb",
                    "original_filename": img_file
                })
                counter += 1

    # Save to CSV
    df = pd.DataFrame(data)
    df.to_csv("Adv_plus_Training_dataset.csv", index=False)
    print(f"Successfully processed {len(df)} images. All images saved")
```

```python
import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator
import os
from tqdm import tqdm
import numpy as np
import pickle # <-- Import pickle for saving state

# ========== State Management Functions (NEW) ==========

def save_training_state(epoch, epsilon_list, val_history, state_path="training_state.pkl"):
    """Saves the current training state to a file."""
    state = {
        'epoch': epoch,
        'epsilon_list': epsilon_list,
        'val_history': val_history,
    }
    with open(state_path, 'wb') as f:
        pickle.dump(state, f)
    print(f"\n---> Training state saved to {state_path}")

def load_training_state(state_path="training_state.pkl"):
    """Loads training state from a file if it exists."""
    if os.path.exists(state_path):
        with open(state_path, 'rb') as f:
            state = pickle.load(f)
        print(f"---> Found previous state. Resuming from epoch {state['epoch']}.")
        return state['epoch'], state['epsilon_list'], state['val_history']
    else:
        print("---> No previous state found. Starting training from scratch.")
        # Return initial default values
        return 0, [], {
            'clean_loss': [], 'clean_acc': [],
            'adv_loss': [], 'adv_acc': []
        }

# ======================================================

# ========== 1. Load GTSRB Dataset Using ImageDataGenerator ==========
# (No changes here, keeping it for context)
data_dir = "GTSRB/Final_Training/Images"
test_dir = "GTSRB/Final_Test/Images"
batch_size = 128
target_size = (240, 240)

train_datagen = ImageDataGenerator(
    rotation_range=15, width_shift_range=0.1, height_shift_range=0.1,
    zoom_range=0.2, brightness_range=[0.8, 1.2], horizontal_flip=False,
    validation_split=0.1
)
val_datagen = ImageDataGenerator(validation_split=0.1)

train_generator = train_datagen.flow_from_directory(
    data_dir, target_size=target_size, batch_size=batch_size,
    class_mode="sparse", subset="training", seed=42
)
val_generator = val_datagen.flow_from_directory(
    data_dir, target_size=target_size, batch_size=batch_size,
    class_mode="sparse", subset="validation", seed=42
)

# ========== 2. FGSM and PGD Attack Functions ==========
# (No changes here)
@tf.function
def fgsm_attack_batch(model, images, labels, epsilon):
    epsilon_scaled = epsilon * 255.0
    with tf.GradientTape() as tape:
        tape.watch(images)
        predictions = model(images, training=True)
        loss = tf.keras.losses.sparse_categorical_crossentropy(labels, predictions)
    gradients = tape.gradient(loss, images)
    adv_images = images + epsilon_scaled * tf.sign(gradients)
    return tf.clip_by_value(adv_images, 0, 255)

@tf.function
def pgd_attack_batch(model, images, labels, epsilon, alpha, iterations):
    epsilon_scaled = epsilon * 255.0
    alpha_scaled = alpha * 255.0
    adv_images = tf.identity(images)
    for _ in range(iterations):
        with tf.GradientTape() as tape:
            tape.watch(adv_images)
            predictions = model(adv_images, training=True)
            loss = tf.keras.losses.sparse_categorical_crossentropy(labels, predictions)
        gradients = tape.gradient(loss, adv_images)
        adv_images = adv_images + alpha_scaled * tf.sign(gradients)
        adv_images = tf.clip_by_value(adv_images, images - epsilon_scaled, images + epsilon_scaled)
        adv_images = tf.clip_by_value(adv_images, 0, 255)
    return adv_images

# ========== 3. Advanced Training Step ==========
# (No changes here)
@tf.function
def adaptive_train_step(model, optimizer, loss_fn, x, y, epsilon_list, alpha_factor, iterations, epoch):
    bs = tf.shape(x)[0]
    base_clean_ratio = 0.75
    final_clean_ratio = 0.5
    progress = tf.cast(epoch, tf.float32) / 35.0
    clean_ratio = base_clean_ratio - (base_clean_ratio - final_clean_ratio) * progress
    clean_portion = tf.cast(tf.math.floor(clean_ratio * tf.cast(bs, tf.float32)), tf.int32)
    adv_portion = bs - clean_portion
    x_clean, y_clean = x[:clean_portion], y[:clean_portion]
    x_adv_raw, y_adv = x[clean_portion:], y[clean_portion:]
    eps_tensor = tf.convert_to_tensor(epsilon_list, dtype=tf.float32)
    eps_idx = tf.random.uniform((), minval=0, maxval=len(epsilon_list), dtype=tf.int32)
    eps = eps_tensor[eps_idx]
    alpha = eps / (iterations * 1.5)
    use_fgsm_prob = 0.7 - (progress * 0.4)
    use_fgsm = tf.random.uniform(()) < use_fgsm_prob
    if use_fgsm:
        x_adv = fgsm_attack_batch(model, x_adv_raw, y_adv, eps)
    else:
        dynamic_iterations = tf.cast(iterations + tf.cast(progress * 5, tf.int32), tf.int32)
        x_adv = pgd_attack_batch(model, x_adv_raw, y_adv, eps, alpha, dynamic_iterations)
    x_all = tf.concat([x_clean, x_adv], axis=0)
    y_all = tf.concat([y_clean, y_adv], axis=0)
    with tf.GradientTape() as tape:
        predictions = model(x_all, training=True)
        loss = loss_fn(y_all, predictions)
    grads = tape.gradient(loss, model.trainable_variables)
    optimizer.apply_gradients(zip(grads, model.trainable_variables))
    acc = tf.keras.metrics.sparse_categorical_accuracy(y_all, predictions)
    return loss, tf.reduce_mean(acc), clean_ratio

# ========== 4. Adversarial Validation Function ==========
# (No changes here)
def adversarial_validation(model, val_generator, loss_fn, epsilon_list, alpha_factor, iterations):
    clean_loss_metric, clean_acc_metric = tf.keras.metrics.Mean(), tf.keras.metrics.SparseCategoricalAccuracy()
    adv_loss_metric, adv_acc_metric = tf.keras.metrics.Mean(), tf.keras.metrics.SparseCategoricalAccuracy()
    print("  Evaluating on clean validation data...")
    for step in tqdm(range(len(val_generator)), desc="Clean Validation", leave=False):
        x_val, y_val = val_generator[step]
        x_val = tf.cast(x_val, tf.float32)
        predictions = model(x_val, training=False)
        clean_loss_metric.update_state(loss_fn(y_val, predictions))
        clean_acc_metric.update_state(y_val, predictions)
    print("  Evaluating on adversarial validation data...")
    for step in tqdm(range(len(val_generator)), desc="Adversarial Validation", leave=False):
        x_val, y_val = val_generator[step]
        x_val = tf.cast(x_val, tf.float32)
        eps = epsilon_list[len(epsilon_list)//2]
        alpha = eps * alpha_factor
        if np.random.random() < 0.5:
            x_adv = fgsm_attack_batch(model, x_val, y_val, eps)
        else:
            x_adv = pgd_attack_batch(model, x_val, y_val, eps, alpha, iterations)
        adv_predictions = model(x_adv, training=False)
        adv_loss_metric.update_state(loss_fn(y_val, adv_predictions))
        adv_acc_metric.update_state(y_val, adv_predictions)
    return (clean_loss_metric.result().numpy(), clean_acc_metric.result().numpy(),
            adv_loss_metric.result().numpy(), adv_acc_metric.result().numpy())

# ========== 5. Train the Model with Adversarial Validation (MODIFIED) ==========

# --- Configuration ---
epochs = 35
iterations = 10
alpha_factor = 1 / iterations
# This is the BASE list of epsilons. It will not be modified.
base_epsilon_list = [0.007, 0.01, 0.03, 0.05, 0.07, 0.1, 0.15]
state_file = "training_state.pkl"

# --- Load State and Model ---
start_epoch, _, val_history = load_training_state(state_file) # Use _ for epsilon_list as we recalculate it

model_path_to_load = f"efficientnet_adv_epoch_{start_epoch}.keras"

if start_epoch > 0 and os.path.exists(model_path_to_load):
    print(f"---> Loading saved model from: {model_path_to_load}")
    model = tf.keras.models.load_model(model_path_to_load)
else:
    print("---> Loading initial base model: EfficientNetB1.keras")
    model = tf.keras.models.load_model("EfficientNetB1.keras")
    # Reset history if we are starting from scratch
    val_history = {'clean_loss': [], 'clean_acc': [], 'adv_loss': [], 'adv_acc': []}

# --- Initialize Optimizer and Loss ---
optimizer = tf.keras.optimizers.Adam(1e-4)
loss_fn = tf.keras.losses.SparseCategoricalCrossentropy()

# --- Main Training Loop (MODIFIED) ---
# The loop now starts from the loaded epoch number
for epoch in range(start_epoch, epochs):
    print(f"\nEpoch {epoch+1}/{epochs}")
    print("=" * 50)

    # Dynamic epsilon scheduling - calculated fresh each epoch from the base list
    # This prevents the bug of repeated multiplication.
    epsilon_list = [e * (1 + epoch / epochs) for e in base_epsilon_list]
    print(f"Epsilon range for this epoch: {[f'{e:.4f}' for e in epsilon_list]}")

    # Training phase
    print("Training...")
    progress_bar = tqdm(range(len(train_generator)), desc=f"Epoch {epoch+1}")
    epoch_losses, epoch_accs = [], []

    for step in progress_bar:
        x_batch, y_batch = train_generator[step]
        x_batch = tf.cast(x_batch, tf.float32)
        loss, acc, clean_ratio = adaptive_train_step(
            model, optimizer, loss_fn, x_batch, y_batch,
            epsilon_list, alpha_factor, iterations, epoch
        )
        epoch_losses.append(loss.numpy())
        epoch_accs.append(acc.numpy())
        progress_bar.set_postfix({
            "Loss": f"{loss:.4f}", "Acc": f"{acc:.4f}", "Clean%": f"{clean_ratio:.0%}"
        })

    print(f"Training Summary — Loss: {np.mean(epoch_losses):.4f}, Accuracy: {np.mean(epoch_accs):.4f}")

    # Validation phase
    print("\nValidation...")
    clean_loss, clean_acc, adv_loss, adv_acc = adversarial_validation(
        model, val_generator, loss_fn, epsilon_list, alpha_factor, iterations
    )

    val_history['clean_loss'].append(clean_loss)
    val_history['clean_acc'].append(clean_acc)
    val_history['adv_loss'].append(adv_loss)
    val_history['adv_acc'].append(adv_acc)

    print(f"\nValidation Summary:")
    print(f"  Clean     — Loss: {clean_loss:.4f}, Accuracy: {clean_acc:.4f}")
    print(f"  Adversarial — Loss: {adv_loss:.4f}, Accuracy: {adv_acc:.4f}")
    print(f"  Robustness Gap: {clean_acc - adv_acc:.4f}")

    # Save checkpoint and state every 5 epochs
    if (epoch + 1) % 3 == 0:
      model_path = f"efficientnet_adv_epoch_{epoch+1}.keras"
      model.save(model_path)
      print(f"\n---> Model checkpoint saved to {model_path}")

      # Copy to Google Drive
      drive_model_path = os.path.join(DRIVE_FOLDER, model_path)
      shutil.copy(model_path, drive_model_path)
      print(f"---> Copied to Google Drive at {drive_model_path}")
      state_path = "training_state.pkl"
      drive_state_path = os.path.join(DRIVE_FOLDER,state_path)


      # Save state for resuming
      save_training_state(epoch + 1, epsilon_list, val_history, state_file)
      shutil.copy(state_file,drive_state_path)

# Save final model
model.save("efficientnet_fgsm_pgd_robust_final.keras")
print("\nFinal model saved to efficientnet_fgsm_pgd_robust_final.keras")

# Final summary using the complete history
print("\n" + "="*60)
print("FINAL TRAINING SUMMARY (Full History)")
print("="*60)
if val_history['clean_acc']:
    print(f"Best Clean Accuracy: {max(val_history['clean_acc']):.4f}")
    print(f"Best Adversarial Accuracy: {max(val_history['adv_acc']):.4f}")
    print(f"Final Clean Accuracy: {val_history['clean_acc'][-1]:.4f}")
    print(f"Final Adversarial Accuracy: {val_history['adv_acc'][-1]:.4f}")
    print(f"Final Robustness Gap: {val_history['clean_acc'][-1] - val_history['adv_acc'][-1]:.4f}")
else:
    print("No validation history recorded.")
```
